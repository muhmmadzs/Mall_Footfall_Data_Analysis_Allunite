Set `FORCE_RECOMPUTE = True` to regenerate CSVs from raw data.

```python
FORCE_RECOMPUTE = False

from notebook_helpers import (
    run_both_plans,
    load_comparison_daily,
    load_mall_visitors_daily,
    load_mall_visitors_hourly,
    plot_comparison_interactive,
    show_figure,
    PLAN_B_OUTPUT,
)
from IPython.display import display

run_both_plans(force_recompute=FORCE_RECOMPUTE)
comp = load_comparison_daily()
mall_daily = load_mall_visitors_daily()

print("Daily comparison (sensor-hour footfall + mall visitors):")
display(comp)
print("\nMall visitors daily:")
display(mall_daily)
```

```python
plot_comparison_interactive(comp)
show_figure(PLAN_B_OUTPUT, "comparison_daily_footfall.png")
```

```python
mh = load_mall_visitors_hourly()
if mh.empty:
    print("Run with FORCE_RECOMPUTE=True if hourly mall file is missing.")
else:
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(13, 4))
    ax.plot(mh["hour_start"], mh["estimated_mall_visitors"], color="#27ae60", linewidth=1.2)
    ax.set_title("Hourly estimated mall visitors")
    ax.set_ylabel("Visitors")
    ax.set_xlabel("Hour")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```
